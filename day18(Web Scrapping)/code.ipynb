```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

all_data = []

for page in range(1, 20):
    url = f"https://www.ambitionbox.com/list-of-companies?page={page}"
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "lxml")

    company_cards = soup.select("div.companyCardWrapper")

    for card in company_cards:
        # Company Name
        name_tag = card.find("h2")
        name = name_tag.text.strip() if name_tag else "NA"

        # Rating
        rating_tag = card.select_one("div.rating_text.rating_text--md > div")
        rating = rating_tag.text.strip() if rating_tag else "NA"

        # Rating Count
        count_tag = card.select_one("span.companyCardWrapper__companyRatingCount")
        rating_count = count_tag.text.strip() if count_tag else "NA"

        # Reviews
        review_tag = card.select_one("a.companyCardWrapper__ActionWrapper")
        review = review_tag.text.strip() if review_tag else "NA"

        # Industry
        industry_loc_tag = card.select_one("span.companyCardWrapper__interLinking")
        industry = (
            industry_loc_tag.text.strip().split("|")[0].strip()
            if industry_loc_tag else "NA"
        
        # Salaries
        salary_data = card.select_one("companyCardWrapper__ActionCount")
        salary = (
            salary_data.text.strip()
            if industry_data else "NA"
        )
        )

        all_data.append({
            "Company Name": name,
            "Rating": rating,
            "Rating Count": rating_count,
            "Reviews": review,
            "Industry": industry
        })

final = pd.DataFrame(all_data)
print(final.head())
```
